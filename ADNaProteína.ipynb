#### Ejercicio 1. Replicación del ADN
Escribe un pequeño script que, dada una cadena de ADN, genere automáticamente su hebra complementaria, y comprueba si tu resultado coincide con lo que obtuviste manualmente.

```python
from Bio.Seq import Seq

original_strand = Seq("ATGCCGTTAGCT")
complementary_strand = original_strand.complement()
manual_result = "TACGGCAATCGA"

print(f"Hebra original (5'->3'): {original_strand}")
print(f"Hebra complementaria (3'->5'): {complementary_strand}")
print(f"Resultado manual (3'->5'): {manual_result}")

if str(complementary_strand) == manual_result:
    print("¡Son iguales!")
else:
    print("El resultado manual es incorrecto")
```


#### Ejercicio 2. Transcripción del ADN a ARN
Crea un script que lea un archivo FASTA con ADN y produzca la secuencia de ARNm. Experimenta cambiando la orientación de la hebra y observa qué ocurre.

```python
from Bio import SeqIO

record = SeqIO.read("lacZ_Ecoli_NC_000913.3.fasta", "fasta")
dna = record.seq

print("ADN original (5' -> 3'):")
print(dna)

mrna = dna.transcribe()

print("\nARNm obtenido:")
print(mrna)

dna_inverted = dna.reverse_complement()

print("\nADN en la orientación contraria (5' -> 3'):")
print(dna_inverted)

mrna_inverted = dna_inverted.transcribe()

print("\nARNm obtenido al cambiar la orientación:")
print(mrna_inverted)

print("\nCOMPARACIÓN")
print("ARNm original:             ", mrna)
print("ARNm orientación contraria:", mrna_inverted)
```

Al cambiar la orientación de la hebra de ADN mediante reverse_complement(), se obtiene una secuencia de ARNm diferente. Esto demuestra que la orientación y la hebra utilizada son importantes durante la transcripción.

#### Ejercicio 3. Traducción del ARNm a proteína

Utiliza el módulo Bio.Seq para traducir automáticamente el ARNm a una secuencia proteica y
comprueba si el resultado coincide con el tuyo.

```python
from Bio.Seq import Seq

mrna = Seq("AUGUAUGCUUAA")

protein = mrna.translate()

manual_result = "MYA*"

print(f"Secuencia de ARNm: {mrna}")
print(f"Secuencia proteica: {protein}")
print(f"Resultado manual: {manual_result}")

if str(protein) == manual_result:
    print("¡Los resultados coinciden!")
else:
    print("Los resultados no coinciden.")
```

#### Ejercicio 6. Actividad integradora: del ADN a la proteína


```python
from Bio import SeqIO

record = SeqIO.read("lacZ_Ecoli_NC_000913.3.fasta", "fasta")
dna = record.seq

print("Iniciamos pipeline")

print("\nSecuencia de ADN cargada correctamente")
print(f"Longitud de la secuencia: {len(dna)} nucleótidos")
print(f"Hebra original (5'->3'): {dna}")


print("\nPASO 1: REPLICACIÓN")
print("Generando la hebra complementaria...")

complementary_strand = dna.complement()

print(f"Hebra original (5'->3'):       {dna}")
print(f"Hebra complementaria (3'->5'): {complementary_strand}")

print("\nLa replicación se ha realizado correctamente")


print("\nPASO 2: TRANSCRIPCIÓN")
print("Transcribiendo el ADN a ARNm...")

mrna = dna.transcribe()

print(f"ADN codificante (5'->3'): {dna}")
print(f"ARNm obtenido (5'->3'):   {mrna}")

print("\nLa transcripción se ha realizado correctamente")


print("\nPASO 3: TRADUCCIÓN")
print("Traduciendo el ARNm a una secuencia de aminoácidos...")

protein = mrna.translate()

print(f"ARNm: {mrna}")
print(f"Proteína obtenida: {protein}")

print("\nLa traducción se ha realizado correctamente")

print("\nPipeline finalizado")
```
